```python
import pandas as pd
```

```python
df = pd.read_csv('winemag-data-130k-v2.csv', index_col=0)
df
```

```python
df.columns = df.columns.str.lower().str.strip()
df
```

```python
df = df.dropna(subset = ['price', 'points'])
df
```

```python
data_final = df[(df['country'] == 'Portugal') & (df['points'] > 90)].copy()
data_final
```

```python
data_final['custo_por_ponto'] = data_final['price'] / data_final['points'] 
data_final
```
